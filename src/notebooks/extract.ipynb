# Extract Step

This notebook reads raw sales data from a CSV file stored in a Unity Catalog volume.

**Source**: `/Volumes/workspace/default/etl_source_data/sales_data.csv`

**Output**: A Spark DataFrame `raw_sales_df` available for downstream transformation.

```python
# Databricks notebook source
# --- Extract: Read raw sales data from UC volume ---

# Define the source path (can be overridden by job parameters)
_widgets = dbutils.widgets.getAll()
SOURCE_PATH = _widgets.get("source_path", "/Volumes/workspace/default/etl_source_data/sales_data.csv")

print(f"Reading CSV from: {SOURCE_PATH}")

# Read the CSV file with header and schema inference
raw_sales_df = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv(SOURCE_PATH)
)

# Display basic stats
row_count = raw_sales_df.count()
print(f"Extracted {row_count} rows")
print(f"Schema:")
raw_sales_df.printSchema()

# Show sample data
raw_sales_df.show(5, truncate=False)

# Make available to downstream notebooks via temp view
raw_sales_df.createOrReplaceTempView("raw_sales")

print("Extract step completed successfully.")
```
